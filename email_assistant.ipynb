```python
from google import genai
from google.genai import types
from google.colab import userdata

client = genai.Client(api_key=userdata.get("GEMINI"))
MODEL = "gemini-3.8-flash"

SYSTEM = "You are an email assistant for sales/support. Output only the email body."

def draft_email(points, tone="professional", language="Arabic"):
    response = client.models.generate_content(
        model=MODEL,
        contents=f"Write a {tone} email in {language} from these points:\n{points}",
        config=types.GenerateContentConfig(
            system_instruction=SYSTEM,
            temperature=0.7,
        ),
    )
    return response.text

print(draft_email("""
- شكر العميل على الاستفسار
- عرض سعر الباقة الأساسية
- اقتراح مكالمة الأسبوع الجاي
"""))
```
